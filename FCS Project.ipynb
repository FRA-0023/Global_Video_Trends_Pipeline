### 1. Create a single dataframe with the concatenation of all input csv files, adding a column called `country`

```python
import pandas as pd
import zstandard
import io
```

```python
countries = ["CA", "DE", "FR", "GB", "IN", "JP", "KR", "MX", "RU", "US"]
dfs = {}

for element in countries:
    path = fr"FCS/{element}videos.csv.zst"
    
    try:
        with open(path, "rb") as f: #binary reading to decompress
            dctx = zstandard.ZstdDecompressor() #object to decompress
            decompressed = dctx.decompress(f.read()) #read the decompressed file
            text = decompressed.decode("utf-8", errors="replace") #transform the decompress file in a readable format for humans (CSV) and cover errors
            dfs[element] = pd.read_csv(io.StringIO(text)) #we expect a csv file not a string to read it, so we convert the string in a csv
            
            dfs[element]['trending_date'] = pd.to_datetime(dfs[element]['trending_date'],format='%y.%d.%m')
            dfs[element]['publish_time'] = pd.to_datetime(dfs[element]['publish_time'])
            dfs[element]["country"] = element
            
    except Exception as e:
        print(f"✗ Errore durante la decompressione di {element}: {e}")

final_dataframe = pd.concat(dfs.values(), ignore_index=True)
final_dataframe
```


### 7. For each interval, determine the number of videos, average number of likes and of dislikes.


```python
#create a function to save the interval number in a string
def cluster_to_time(cluster):
    minutes = cluster * 10
    hours = minutes // 60
    remaining_minutes = minutes % 60
    start_time = f"{hours:02d}:{remaining_minutes:02d}" #format integer with 2 figures for each hours and minutes
    end_time = f"{(minutes + 10) // 60:02d}:{(minutes + 10) % 60:02d}" #calculate the interval by the minutes numbers
    return f"{start_time} - {end_time}" 

avg = final_dataframe[['likes','dislikes']].groupby(final_dataframe['cluster']).mean().round(0)
suum = final_dataframe['video_id'].groupby(final_dataframe['cluster']).count()
clu = avg.join(suum.rename('count'))
clu.loc[:, 'interval'] = clu.index.map(cluster_to_time)
clu
```
